```python
import geopandas as gpd
from pyrosm import OSM
```

```python
# Point to your local .osm.pbf file
pbf_path = "/data/uscuni-ulce/extension2/nor/norway-261020.osm.pbf"
osm = OSM(pbf_path)

# Extract a specific layer, e.g., buildings, road network, or POIs
gdf = osm.get_buildings()
```

```python
gdf["area"] = gdf.geometry.area          
top10 = gdf.nlargest(100, "area")
```

```python
filt = gdf[gdf["building"] != "no"]
```

```python
filt = filt.to_crs(3035)
```

```python
filt.to_parquet("/data/uscuni-ulce/extension2/nor/clean_3035.parquet")
```
